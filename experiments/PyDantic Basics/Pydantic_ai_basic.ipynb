```python
from pydantic import BaseModel, Field, field_validator
```

```python
class UserSignUp(BaseModel):
    username : str = Field(min_length=3, max_length=15)
    age : int = Field(ge=13)
    account_type : str = "free"
    
    @field_validator('username')
    @classmethod
    def check_user_no_spaces(cls, value: str) -> str:
        if " " in value:
            raise ValueError("Usernames cannot contains spaces!")
        return value
```

```python
user1 = UserSignUp(
    username="gyan",
    age=14
)
print(user1)
```

```python
bad_username = UserSignUp(
    username= "cool guy",
    age = 15
)
print(bad_username)
```
